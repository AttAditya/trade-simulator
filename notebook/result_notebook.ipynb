## Visual Analysis

```python
traders = {trader for trader in {*data["trader_name"]} if trader[0] != "[" }
```

```python
def plot_traders_field(traders, field):
  for trader in traders:
    trader_data = data[data['trader_name'] == trader][['completion', field]]
    plt.plot(trader_data['completion'], trader_data[field], label=trader)
  
  plt.xlabel('Completion')
  plt.ylabel(field)
  
  plt.title(f'{field} vs Completion for each Trader')
  plt.legend([trader for trader in traders])
  plt.show()
```

```python
def plot_traders_normalized_field(traders, field):
  for trader in traders:
    trader_data = data[data['trader_name'] == trader][['completion', field]]
    plt.plot(
      trader_data['completion'],
      (
        (trader_data[field] - trader_data[field].min()) /
        (trader_data[field].max() - trader_data[field].min())
      ),
      label=trader
    )
  
  plt.xlabel('Completion')
  plt.ylabel(field)
  
  plt.title(f'{field} vs Completion for each Trader')
  plt.legend([trader for trader in traders])
  plt.show()
```

```python
plot_traders_field(traders, "total_wealth")
```

```python
plot_traders_normalized_field(traders, "total_wealth")
```

```python
plot_traders_field({
  "Momentum Monkey V3",
  "Momentum Monkey V2",
}, "total_wealth")
```

```python
plot_traders_field(traders, "cash")
```

```python
plot_traders_field(traders, "debit_count")
```

```python
plot_traders_field(traders, "credit_count")
```
